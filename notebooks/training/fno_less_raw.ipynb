## Импорты и путь к проекту


```python
import numpy as np
import sys
import os
import time
import random
import torch
from torch.optim.lr_scheduler import CosineAnnealingLR
from torch.utils.data import DataLoader


from galaxy_mass_estimation.models.fno import ModelFNO2
from galaxy_mass_estimation.data import ACTClusterDataset
```

## Воспроизводимость


```python
def set_seed(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

    os.environ["PYTHONHASHSEED"] = str(seed)


set_seed(42)
```

## Устройство


```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

## Параметры обучения


```python
EPOCHS = 100
LR = 1e-3
BATCH_SIZE = 128
```

## Модель, оптимизатор и функция потерь


```python
model = ModelFNO2().to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
scheduler = CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-5)
loss_fn = torch.nn.MSELoss()
```

## Индексы выборок


```python
train_indices = np.load(PROJECT_ROOT / "data/splits/raw/train_indices.npy")
test_indices = np.load(PROJECT_ROOT / "data/splits/raw/test_indices.npy")
```

## Датасеты


```python
train_dataset = ACTClusterDataset(str(PROJECT_ROOT / "data/processed"), train_indices)
test_dataset = ACTClusterDataset(str(PROJECT_ROOT / "data/processed"), test_indices)
```

## Загрузчики данных


```python
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=8,
    pin_memory=True,
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=8,
    pin_memory=True,
)
```

## Обучение и промежуточная оценка


```python
model.train()
losses = []

for epoch in range(EPOCHS):
    start_t = time.time()
    total_loss = 0
    for img, scalars, target in train_loader:
        img = img.to(device)
        scalars = scalars.to(device)
        target = target.to(device)

        optimizer.zero_grad()
        pred = model(img, scalars)
        loss = loss_fn(pred, target)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    losses.append(total_loss / len(train_loader))

    scheduler.step()

    if epoch == 0:
        print("Epoch 1 time:", time.time() - start_t)

    # --- оценка каждые 10 эпох ---
    if (epoch + 1) % 10 == 0:
        model.eval()
        preds, targets = [], []
        with torch.no_grad():
            for img, scalars, target in test_loader:
                img = img.to(device)
                scalars = scalars.to(device)
                target = target.to(device)

                pred = model(img, scalars)
                preds.append(pred)
                targets.append(target)

        model.train()

        preds = torch.cat(preds)
        targets = torch.cat(targets)

        mae = torch.abs(targets - preds)
        mean_mae = mae.mean().item()
        max_mae = mae.max().item()

        relative_errors = mae / targets
        mean_rel_error = relative_errors.mean().item()
        max_rel_error = relative_errors.max().item()

        print(f"EPOCH={epoch+1}")
        print(f"Средняя MAE {mean_mae:.4f}")
        print(f"Максимальная MAE {max_mae:.4f}")
        print(f"Средняя относительная ошибка: {mean_rel_error:.4f}")
        print(f"Максимальная относительная ошибка: {max_rel_error:.4f}")
losses = np.asarray(losses)
```

## Итоговые предсказания


```python
model.eval()  # режим оценки
preds = []
targets = []

with torch.no_grad():
    for img, scalars, target in test_loader:
        img = img.to(device)
        scalars = scalars.to(device)
        target = target.to(device)

        pred = model(img, scalars)

        preds.append(pred.cpu())
        targets.append(target.cpu())

preds = torch.cat(preds)
targets = torch.cat(targets)

mae = torch.abs(targets - preds)
mean_mae = mae.mean().item()
max_mae = mae.max().item()

relative_errors = mae / torch.abs(targets)
mean_rel_error = relative_errors.mean().item()
max_rel_error = relative_errors.max().item()

print(f"EPOCH={epoch+1}")
print(f"Средняя MAE: {mean_mae:.4f}")
print(f"Максимальная MAE: {max_mae:.4f}")
print(f"Средняя относительная ошибка: {mean_rel_error:.4f}")
print(f"Максимальная относительная ошибка: {max_rel_error:.4f}")
```

## Сохранение результатов


```python
output_dir = PROJECT_ROOT / "results/fno_less/raw"
output_dir.mkdir(parents=True, exist_ok=True)
np.save(output_dir / "losses.npy", losses)
np.save(output_dir / "targets.npy", targets)
np.save(output_dir / "predictions.npy", preds)
```
