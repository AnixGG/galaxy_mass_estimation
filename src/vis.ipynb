```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
targets = np.load("results/base/base/base_base_targets.npy")
preds = np.load("results/base/base/base_base_preds.npy")
```

```python
def show_vis(targets, preds, title):
    # Преобразуем в numpy и выравниваем по длине
    targets = np.array(targets).flatten()
    preds = np.array(preds).flatten()

    # Проверим длины
    n = min(len(targets), len(preds))
    targets, preds = targets[:n], preds[:n]

    # Построение графика
    plt.figure(figsize=(10, 5))
    plt.plot(range(n), targets, label="Targets", linewidth=2)
    plt.plot(range(n), preds, label=f"Predictions ({title})", linewidth=2, alpha=0.8)

    # Оформление
    plt.xlabel("Sample", fontsize=12)
    plt.ylabel("Mass", fontsize=12)
    plt.title(f"Сравнение Targets и Predictions ({title})", fontsize=14, pad=10)
    plt.legend(frameon=False)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(5,5))
    plt.scatter(targets, preds, s=10, alpha=0.6)
    plt.plot([targets.min(), targets.max()],
             [targets.min(), targets.max()],
             'r--', lw=2, label="идеал")
    plt.xlabel("Target")
    plt.ylabel("Prediction")
    plt.title(f"Диагональное сравнение ({title})")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()
    
    mae = abs(targets - preds)
    max_mae = mae.max().item()
    mean_mae = mae.mean().item()

    rel_error = mae / targets
    max_rel_error = rel_error.max().item() * 100
    mean_rel_error = rel_error.mean().item() * 100
    
    mse = np.square(mae)
    max_mse = mse.max().item()
    mean_mse = mse.mean().item()

    print(f"Средняя MAE: {mean_mae:.4f}")
    print(f"Максимальная MAE: {max_mae:.4f}")
    
    print(f"Средняя MSE: {mean_mse:.4f}")
    print(f"Максимальная MSE: {max_mse:.4f}")
    
    print(f"Средняя относительная ошибка: {mean_rel_error:.4f}%")
    print(f"Максимальная относительная ошибка: {max_rel_error:.4f}%")
    
```

```python
show_vis(targets, preds, "fno depth=4")
```
